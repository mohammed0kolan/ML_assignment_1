```python
import pandas as pd

df = pd.read_csv("Mobile-data - Mobile-data.csv")
df.head()
```

```python
print("Rows:", df.shape[0], ", columns:",df.shape[1] )
print("\nColumn names:")
print(df.columns.tolist())
```

```python
numeric_df = df.select_dtypes(include="number")

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1

outlier_mask = (numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))

outlier_mask.sum().sort_values(ascending=False)
```

```python
for column in ["three_g", "fc", "px_height", "px_width", "battery_power"]:
    print(f"\n{column}:")
    print(df[column].value_counts().sort_index().to_string())
```

```python
df.loc[
    (df["battery_power"] < 0) | (df["px_width"] < 0),
    ["battery_power", "px_width", "px_height", "fc", "price_range"]
]
df.loc[[0, 65]].T
```

```python
before = len(df)

invalid_mask = (df["battery_power"] < 0) | (df["px_width"] < 0)
df_clean = df.loc[~invalid_mask].copy()

after = len(df_clean)

print("Rows before:", before)
print("Invalid rows removed:", invalid_mask.sum())
print("Rows after:", after)
```

```python
df_clean.isnull().sum()
```

```python
df_clean.isnull().sum().sum()
```

```python
df_clean["screen_area"] = df_clean["sc_w"] * df_clean["sc_h"]

correlations = df_clean[["sc_w", "sc_h", "screen_area", "price_range"]].corr()
correlations["price_range"].drop("price_range").sort_values(ascending=False)
```

```python
from sklearn.preprocessing import MinMaxScaler

feature_cols = df_clean.columns.drop("price_range")

scaler = MinMaxScaler()
df_scaled = df_clean.copy()
df_scaled[feature_cols] = scaler.fit_transform(df_clean[feature_cols])

# Compare original and scaled values
print("Before scaling:")
display(df_clean[["battery_power", "ram", "price_range"]].head())

print("After scaling:")
display(df_scaled[["battery_power", "ram", "price_range"]].head())

# Check each scaled feature's range
df_scaled[feature_cols].agg(["min", "max"]).T
```

```python


import seaborn as sns
import matplotlib.pyplot as plt

ax = sns.countplot(data=df_clean, x="price_range")

for bar in ax.patches:
    count = int(bar.get_height())
    ax.annotate(
        str(count),
        (bar.get_x() + bar.get_width() / 2, count),
        ha="center",
        va="bottom"
    )

ax.set_title("Number of Phones in Each Price Range")
ax.set_xlabel("Price range")
ax.set_ylabel("Number of phones")
plt.show()
```

```python
binary_features = ["dual_sim", "blue", "touch_screen", "wifi"]

for feature in binary_features:
    print(f"\n{feature} — counts")
    print(pd.crosstab(df_clean["price_range"], df_clean[feature]))

    print(f"\n{feature} — percentage within each price range")
    print(
        pd.crosstab(
            df_clean["price_range"],
            df_clean[feature],
            normalize="index"
        ).mul(100).round(1)
    )
```

```python
for feature in ["dual_sim", "blue", "touch_screen", "wifi"]:
    print(f"\n{feature} (% within price range)")
    percentages = pd.crosstab(
        df_clean["price_range"],
        df_clean[feature],
        normalize="index"
    ).mul(100).round(1)
    print(percentages.to_string())
```

```python
import numpy as np

corr_matrix = df_clean.corr(numeric_only=True)

# List each feature pair once, without the diagonal
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)
pairs = upper_triangle.stack().sort_values()

print("Most negative correlations:")
print(pairs.head(10))

print("\nMost positive correlations:")
print(pairs.tail(10).sort_values(ascending=False))

print("\nCorrelations with price_range:")
print(
    corr_matrix["price_range"]
    .drop("price_range")
    .sort_values(ascending=False)
)
```

```python
pairs = upper_triangle.stack().dropna().sort_values()

print("Most positive correlations:")
print(pairs.tail(10).sort_values(ascending=False))
```

```python
for column in ["fc", "px_height", "ram"]:
    q1 = df_clean[column].quantile(0.25)
    q3 = df_clean[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    flagged = df_clean.loc[
        (df_clean[column] < lower) | (df_clean[column] > upper),
        [column, "price_range"]
    ]

    print(f"\n{column}: IQR limits {lower:.2f} to {upper:.2f}")
    print(flagged.to_string())
    
```

```python
rows_before_ram_check = len(df_clean)

ram_upper = df_clean["ram"].quantile(0.75) + 1.5 * (
    df_clean["ram"].quantile(0.75) - df_clean["ram"].quantile(0.25)
)

df_final = df_clean.loc[df_clean["ram"] <= ram_upper].copy()

print("Rows before RAM cleanup:", rows_before_ram_check)
print("Rows removed:", rows_before_ram_check - len(df_final))
print("Rows after:", len(df_final))
```

```python
df_final["price_range"].value_counts().sort_index()
```

```python
from sklearn.preprocessing import MinMaxScaler

feature_cols = df_final.columns.drop("price_range")

scaler = MinMaxScaler()
df_scaled_final = df_final.copy()
df_scaled_final[feature_cols] = scaler.fit_transform(df_final[feature_cols])

df_scaled_final[feature_cols].agg(["min", "max"]).T
```

```python
for feature in ["dual_sim", "blue", "touch_screen", "wifi"]:
    print(f"\n{feature} (% within price range)")
    percentages = pd.crosstab(
        df_final["price_range"],
        df_final[feature],
        normalize="index"
    ).mul(100).round(1)
    print(percentages.to_string())
```
